```python
# %%
import glob, os
import numpy as np
import matplotlib.pyplot as plt

EPISODES_DIR = os.path.expanduser("~/dissertation/context-aware-magail-grf/data/demonstrations/episodes")

paths = sorted(glob.glob(os.path.join(EPISODES_DIR, "*.npz")))
print(f"Found {len(paths)} episode files:")
for p in paths:
    print(" ", os.path.basename(p))

# %%
episodes = {}
for p in paths:
    d = dict(np.load(p, allow_pickle=True))
    episodes[os.path.basename(p)] = d

print(f"Loaded {len(episodes)} episodes.")
example = next(iter(episodes.values()))
print("\nFields in each episode:")
for k, v in example.items():
    shape = getattr(v, 'shape', None)
    print(f"  {k:25s} shape={shape}  dtype={getattr(v, 'dtype', type(v))}")

# %%
REGIONS = ['early_neutral', 'mid_neutral', 'late_winning', 'late_losing', 'other']

def classify_region(t_norm, delta_score):
    if t_norm > 0.7 and delta_score == 0:
        return 'early_neutral'
    if 0.4 <= t_norm <= 0.6 and delta_score == 0:
        return 'mid_neutral'
    if t_norm < 0.3 and delta_score > 0:
        return 'late_winning'
    if t_norm < 0.3 and delta_score < 0:
        return 'late_losing'
    return 'other'

def compute_sap(sticky_actions):
    return float(sticky_actions[:, 8].mean() * 100)

rows = []
for name, d in episodes.items():
    t_norm = d['steps_left'] / 3001.0
    delta_score = d['score_left'] - d['score_right']
    region = np.array([classify_region(t, ds) for t, ds in zip(t_norm, delta_score)])
    region_counts = {r: int((region == r).sum()) for r in REGIONS}
    sap = compute_sap(d['sticky_actions'])
    poss = float(d['has_possession'].mean())
    rows.append({
        'file': name,
        'episode_id': int(d['episode_id']),
        'n_steps': len(d['steps_left']),
        'sap_%': round(sap, 2),
        'possession_%': round(poss * 100, 1),
        **region_counts,
    })

try:
    import pandas as pd
    df = pd.DataFrame(rows).sort_values('episode_id').reset_index(drop=True)
    display(df)
except ImportError:
    for r in sorted(rows, key=lambda x: x['episode_id']):
        print(r)

# %%
totals = {r: 0 for r in REGIONS}
for row in rows:
    for r in REGIONS:
        totals[r] += row[r]

print("Total steps per region across all loaded episodes:")
for r in ['early_neutral', 'mid_neutral', 'late_winning', 'late_losing']:
    flag = "OK" if totals[r] > 0 else "MISSING"
    print(f"  {r:15s}: {totals[r]:6d} steps  [{flag}]")
print(f"  {'other':15s}: {totals['other']:6d} steps")

plt.figure(figsize=(6,4))
labels = ['early_neutral','mid_neutral','late_winning','late_losing','other']
plt.bar(labels, [totals[l] for l in labels])
plt.xticks(rotation=30, ha='right')
plt.ylabel("total steps")
plt.title("Context region coverage, all episodes combined")
plt.tight_layout()
plt.show()

# %%
# --- pick one episode to inspect closely ---
# change EP_NAME to any filename printed in the first cell, then re-run cells below
EP_NAME = paths and os.path.basename(paths[0])
d = episodes[EP_NAME]
print(f"Inspecting: {EP_NAME}")
print(f"episode_id: {d['episode_id']}   context_label: {d['context_label']}")
print(f"n_steps: {len(d['steps_left'])}")

# %%
for k in ['steps_left', 'score_left', 'score_right', 'ball_owned_team',
          'ball_owned_player', 'action_captured', 'has_possession']:
    print(f"{k:20s}: {d[k][:5]}")

print()
print("left_team (player positions), step 0, all 5 players:")
print(d['left_team'][0])
print()
print("sticky_actions, step 0 (8 directions + sprint + dribble):")
print(d['sticky_actions'][0])

# %%
t_norm = d['steps_left'] / 3001.0
delta_score = d['score_left'] - d['score_right']
region = np.array([classify_region(t, ds) for t, ds in zip(t_norm, delta_score)])

region_to_int = {r: i for i, r in enumerate(REGIONS)}
region_ints = np.array([region_to_int[r] for r in region])

plt.figure(figsize=(10, 2.5))
plt.plot(region_ints, drawstyle='steps-post')
plt.yticks(range(len(REGIONS)), REGIONS)
plt.xlabel("step")
plt.title(f"Context region over time -- {EP_NAME}")
plt.tight_layout()
plt.show()

region_counts = {r: int((region == r).sum()) for r in REGIONS}
print("Region counts this episode:", region_counts)

# %%
fig, ax1 = plt.subplots(figsize=(10, 3))
ax1.plot(d['score_left'], label='score_left', color='tab:blue')
ax1.plot(d['score_right'], label='score_right', color='tab:red')
ax1.set_xlabel('step')
ax1.set_ylabel('score')
ax1.legend(loc='upper left')

ax2 = ax1.twinx()
ax2.plot(t_norm, label='T_norm', color='tab:green', alpha=0.5)
ax2.set_ylabel('T_norm (starts at 1.0, decreases)')
ax2.legend(loc='upper right')

plt.title(f"Score & T_norm -- {EP_NAME}")
plt.tight_layout()
plt.show()

# %%
ACTION_NAMES = {
    0: 'IDLE', 1: 'LEFT', 2: 'TOP_LEFT', 3: 'TOP', 4: 'TOP_RIGHT',
    5: 'RIGHT', 6: 'BOTTOM_RIGHT', 7: 'BOTTOM', 8: 'BOTTOM_LEFT',
    9: 'LONG_PASS', 10: 'HIGH_PASS', 11: 'SHORT_PASS', 12: 'SHOT',
    13: 'SPRINT', 14: 'RELEASE_DIRECTION', 15: 'RELEASE_SPRINT',
    16: 'SLIDING', 17: 'DRIBBLE', 18: 'RELEASE_DRIBBLE',
}
counts = np.bincount(d['action_captured'], minlength=19)

plt.figure(figsize=(10, 4))
plt.bar(range(19), counts)
plt.xticks(range(19), [ACTION_NAMES[i] for i in range(19)], rotation=60, ha='right')
plt.ylabel("count")
plt.title(f"Action histogram -- {EP_NAME}")
plt.tight_layout()
plt.show()

sprint_on, sprint_off = counts[13], counts[15]
dribble_on, dribble_off = counts[17], counts[18]
print(f"sprint on/off: {sprint_on}/{sprint_off}  (should differ by at most 1)")
print(f"dribble on/off: {dribble_on}/{dribble_off}  (should differ by at most 1)")

# %%
plt.figure(figsize=(8, 5))
colors = plt.cm.tab10(np.linspace(0, 1, 5))
for i in range(5):
    xs = d['left_team'][:, i, 0]
    ys = d['left_team'][:, i, 1]
    plt.plot(xs, ys, alpha=0.4, color=colors[i], label=f'player {i}')
plt.plot(d['ball'][:, 0], d['ball'][:, 1], color='black', alpha=0.6, linewidth=0.7, label='ball')
plt.xlabel('pitch x'); plt.ylabel('pitch y')
plt.legend(loc='upper right', fontsize=8)
plt.title(f"Position trails -- {EP_NAME}")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

# %%
# quick scan of all 10 episodes at once, no need to change EP_NAME repeatedly
for name, dd in sorted(episodes.items(), key=lambda kv: int(kv[1]['episode_id'])):
    ac = np.bincount(dd['action_captured'], minlength=19)
    sprint_on, sprint_off = ac[13], ac[15]
    dribble_on, dribble_off = ac[17], ac[18]
    sap = compute_sap(dd['sticky_actions'])
    flags = []
    if abs(int(sprint_on) - int(sprint_off)) > 1:
        flags.append("sprint mismatch")
    if abs(int(dribble_on) - int(dribble_off)) > 1:
        flags.append("dribble mismatch")
    if sap > 40:
        flags.append("SAP exceeds 40% gate")
    print(f"{name:30s} SAP={sap:5.2f}%  sprint={sprint_on}/{sprint_off}  dribble={dribble_on}/{dribble_off}"
          + (f"  FLAGS: {flags}" if flags else ""))
```

```python
# %% [2] Flatten every episode into a full step-level DataFrame (all raw features as columns)
import pandas as pd

STICKY_NAMES = ['sticky_left', 'sticky_top_left', 'sticky_top', 'sticky_top_right',
                 'sticky_right', 'sticky_bottom_right', 'sticky_bottom', 'sticky_bottom_left',
                 'sticky_sprint', 'sticky_dribble']

def flatten_episode(name, d):
    n = len(d['steps_left'])
    cols = {}

    cols['episode_id'] = np.full(n, int(d['episode_id']))
    cols['context_label'] = np.full(n, str(d['context_label']))
    cols['source_file'] = np.full(n, name)
    cols['step'] = np.arange(n)

    # scalar-per-step fields
    for k in ['steps_left', 'score_left', 'score_right', 'ball_owned_team',
              'ball_owned_player', 'action_captured', 'has_possession',
              'context_region']:
        cols[k] = d[k]

    # (T, 5, 2) fields -- team positions/directions
    for field in ['left_team', 'left_team_direction', 'right_team', 'right_team_direction']:
        for p in range(5):
            cols[f'{field}_p{p}_x'] = d[field][:, p, 0]
            cols[f'{field}_p{p}_y'] = d[field][:, p, 1]

    # (T, 3) fields -- ball position/direction
    for field in ['ball', 'ball_direction']:
        for i, axis in enumerate(['x', 'y', 'z']):
            cols[f'{field}_{axis}'] = d[field][:, i]

    # (T, 10) sticky_actions
    for i, sname in enumerate(STICKY_NAMES):
        cols[sname] = d['sticky_actions'][:, i]

    return pd.DataFrame(cols)


# one DataFrame per episode, kept separately (each is n_steps rows x all-feature columns)
episode_dfs = {name: flatten_episode(name, d) for name, d in episodes.items()}

# also build one big concatenated DataFrame across all 52 episodes, for full-dataset analysis
all_df = pd.concat(episode_dfs.values(), ignore_index=True)

print(f"episode_dfs: {len(episode_dfs)} DataFrames, one per episode")
example_name = next(iter(episode_dfs))
print(f"  example ({example_name}): shape = {episode_dfs[example_name].shape}")
print(f"\nall_df (all episodes concatenated): shape = {all_df.shape}")
print(f"\ncolumns ({len(all_df.columns)} total):")
print(list(all_df.columns))

all_df.head()
```

```python
# %% [3] SAP / MECHA per episode, then aggregate across all 52
import sys, os
sys.path.insert(0, os.path.expanduser("~/dissertation/context-aware-magail-grf/src/metrics"))
from calculator import compute_sap, compute_mecha

per_episode = []
for name, d in episodes.items():
    sprint_arr = d['sticky_actions'][:, 8:9]        # sprint bit, per-step
    pos_arr = d['left_team'][:, :5]                  # (T, 5, 2) positions
    poss_arr = d['has_possession']                   # (T,) bool

    sap = compute_sap(sprint_arr)
    mecha = compute_mecha(pos_arr, poss_arr)

    per_episode.append({
        'source_file': name,
        'episode_id': int(d['episode_id']),
        'sap': sap,
        'mecha': mecha,
    })

metrics_df = pd.DataFrame(per_episode).sort_values('episode_id').reset_index(drop=True)

print(f"Across all {len(metrics_df)} episodes:\n")
print(f"SAP    -- avg: {metrics_df['sap'].mean():.2f}%   "
      f"highest: {metrics_df['sap'].max():.2f}% ({metrics_df.loc[metrics_df['sap'].idxmax(), 'source_file']})   "
      f"lowest: {metrics_df['sap'].min():.2f}% ({metrics_df.loc[metrics_df['sap'].idxmin(), 'source_file']})")

print(f"MECHA  -- avg: {metrics_df['mecha'].mean():.4f}   "
      f"highest: {metrics_df['mecha'].max():.4f} ({metrics_df.loc[metrics_df['mecha'].idxmax(), 'source_file']})   "
      f"lowest: {metrics_df['mecha'].min():.4f} ({metrics_df.loc[metrics_df['mecha'].idxmin(), 'source_file']})")

metrics_df
```

```python
# %% [4] Win / draw / loss per episode, based on final score, + bar chart
import matplotlib.pyplot as plt

def final_result(d):
    final_left = int(d['score_left'][-1])
    final_right = int(d['score_right'][-1])
    if final_left > final_right:
        return 'win'
    if final_left < final_right:
        return 'loss'
    return 'draw'

results = []
for name, d in episodes.items():
    results.append({
        'source_file': name,
        'episode_id': int(d['episode_id']),
        'final_score_left': int(d['score_left'][-1]),
        'final_score_right': int(d['score_right'][-1]),
        'result': final_result(d),
    })

results_df = pd.DataFrame(results).sort_values('episode_id').reset_index(drop=True)

counts = results_df['result'].value_counts().reindex(['win', 'draw', 'loss'], fill_value=0)
print("Result counts across all episodes:")
print(counts)
print()
print(f"Total: {counts.sum()} episodes")

plt.figure(figsize=(5, 4))
plt.bar(counts.index, counts.values, color=['tab:green', 'tab:gray', 'tab:red'])
for i, v in enumerate(counts.values):
    plt.text(i, v + 0.3, str(v), ha='center')
plt.ylabel('episode count')
plt.title(f'Win / Draw / Loss across {counts.sum()} demonstration episodes')
plt.tight_layout()
plt.show()

results_df
```

```python
# %% [6] Per-step win/draw/loss zone classification (3001 x 52 steps), MECHA stats per zone
from scipy.spatial import ConvexHull
from scipy.spatial.qhull import QhullError

A_PITCH = 2.0 * 0.84  # x in [-1,1], y in [-0.42,0.42] -- matches your earlier position-trail plot

def step_hull_area(positions_5x2):
    """Instantaneous convex hull area for one step's 5 player positions, pitch-normalized."""
    try:
        hull = ConvexHull(positions_5x2)
        return hull.volume / A_PITCH  # 'volume' is area in 2D for scipy's ConvexHull
    except QhullError:
        return np.nan  # degenerate config (e.g. collinear players) -- rare, excluded from stats

# build per-step zone + hull area across ALL episodes, all steps
records = []
for name, d in episodes.items():
    n = len(d['steps_left'])
    delta_score = d['score_left'] - d['score_right']
    zone = np.where(delta_score > 0, 'win', np.where(delta_score < 0, 'loss', 'draw'))
    sprint = d['sticky_actions'][:, 8]
    poss = d['has_possession']

    hull_areas = np.full(n, np.nan)
    for t in range(n):
        if poss[t]:  # MECHA is possession-restricted -- match that here
            hull_areas[t] = step_hull_area(d['left_team'][t])

    for t in range(n):
        records.append({
            'source_file': name, 'step': t, 'zone': zone[t],
            'sprint': sprint[t], 'has_possession': poss[t], 'hull_area': hull_areas[t],
        })

steps_df = pd.DataFrame(records)
print(f"Total steps: {len(steps_df)}  (expect 52 x 3001 = {52*3001})")
print(steps_df['zone'].value_counts())

# --- sanity check: per-step hull area, averaged over one episode's possession steps,
#     should match calculator.py's compute_mecha for that same episode ---
check_name = next(iter(episodes))
check_df = steps_df[(steps_df['source_file'] == check_name) & (steps_df['has_possession'])]
manual_mecha = check_df['hull_area'].mean()
official_mecha = compute_mecha(episodes[check_name]['left_team'][:, :5], episodes[check_name]['has_possession'])
print(f"\nSanity check on {check_name}: manual mean={manual_mecha:.4f}  vs  compute_mecha={official_mecha:.4f}")
print("(should match closely -- if not, the per-step hull logic disagrees with the canonical metric)")

# --- SAP per zone (all steps, matches SAP's own definition -- not possession-restricted) ---
sap_by_zone = steps_df.groupby('zone')['sprint'].mean().reindex(['win', 'draw', 'loss']) * 100

# --- MECHA (hull area) per zone, possession steps only, mean/std/min/max ---
mecha_by_zone = steps_df[steps_df['has_possession']].groupby('zone')['hull_area'].agg(
    ['mean', 'std', 'min', 'max', 'count']
).reindex(['win', 'draw', 'loss'])

print("\nSAP by zone (%):")
print(sap_by_zone)
print("\nMECHA (hull area) by zone, possession steps only:")
print(mecha_by_zone)

# --- visualizations ---
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].bar(sap_by_zone.index, sap_by_zone.values, color=['tab:green', 'tab:gray', 'tab:red'])
axes[0].set_ylabel('SAP (%)')
axes[0].set_title('Sprint Action % by score zone')

means = mecha_by_zone['mean'].values
stds = mecha_by_zone['std'].values
axes[1].bar(mecha_by_zone.index, means, yerr=stds, capsize=6, color=['tab:green', 'tab:gray', 'tab:red'])
axes[1].set_ylabel('MECHA (hull area, mean ± std)')
axes[1].set_title('Formation area by score zone (possession steps)')

plt.tight_layout()
plt.show()

# distribution view -- boxplot, shows range/spread rather than just mean/std
plt.figure(figsize=(6, 4))
data_for_box = [steps_df[(steps_df['zone']==z) & (steps_df['has_possession'])]['hull_area'].dropna()
                for z in ['win', 'draw', 'loss']]
plt.boxplot(data_for_box, labels=['win', 'draw', 'loss'])
plt.ylabel('hull area (MECHA)')
plt.title('MECHA distribution by score zone (possession steps)')
plt.tight_layout()
plt.show()
```

```python
# %% [7] Add time-zone (early/mid/late) to each step, build the 3x3 win/draw/loss x early/mid/late grid
# Convention: whole episode treated as a 90-minute-equivalent match, split 20/50/20 --
# proportional to elapsed steps, not literal GRF wall-clock (flagging this assumption
# explicitly since GRF's own step-to-real-time rate isn't something we've verified here).

def classify_time_zone(elapsed_min):
    if elapsed_min < 20:
        return 'early'
    if elapsed_min < 70:  # 20 + 50
        return 'mid'
    return 'late'

time_records = []
for name, d in episodes.items():
    n = len(d['steps_left'])
    t_norm = d['steps_left'] / 3001.0
    elapsed_min = (1 - t_norm) * 90.0
    tz = np.array([classify_time_zone(m) for m in elapsed_min])
    for t in range(n):
        time_records.append({'source_file': name, 'step': t, 'elapsed_min': elapsed_min[t], 'time_zone': tz[t]})

time_zone_df = pd.DataFrame(time_records)

# merge onto steps_df from the previous cell (must have run [6] first)
steps_df = steps_df.merge(time_zone_df, on=['source_file', 'step'], how='left')

print(steps_df['time_zone'].value_counts())
print()
print("Combined 3x3 grid step counts (time_zone x zone):")
grid_counts = steps_df.groupby(['time_zone', 'zone']).size().unstack(fill_value=0)
grid_counts = grid_counts.reindex(index=['early', 'mid', 'late'], columns=['win', 'draw', 'loss'])
print(grid_counts)

# --- SAP and MECHA per (time_zone, zone) cell ---
sap_grid = steps_df.groupby(['time_zone', 'zone'])['sprint'].mean().unstack() * 100
sap_grid = sap_grid.reindex(index=['early', 'mid', 'late'], columns=['win', 'draw', 'loss'])

mecha_grid_stats = steps_df[steps_df['has_possession']].groupby(['time_zone', 'zone'])['hull_area'].agg(
    ['mean', 'std', 'count']
)
mecha_mean_grid = mecha_grid_stats['mean'].unstack().reindex(index=['early','mid','late'], columns=['win','draw','loss'])
mecha_std_grid = mecha_grid_stats['std'].unstack().reindex(index=['early','mid','late'], columns=['win','draw','loss'])

print("\nSAP (%) grid:")
print(sap_grid.round(2))
print("\nMECHA mean grid:")
print(mecha_mean_grid.round(4))
print("\nMECHA std grid:")
print(mecha_std_grid.round(4))

# --- visualizations: heatmaps for step counts, SAP, MECHA mean ---
fig, axes = plt.subplots(1, 3, figsize=(16, 4))

for ax, data, title, fmt in [
    (axes[0], grid_counts, 'Step count', 'd'),
    (axes[1], sap_grid, 'SAP (%)', '.1f'),
    (axes[2], mecha_mean_grid, 'MECHA mean', '.3f'),
]:
    im = ax.imshow(data.values, cmap='viridis', aspect='auto')
    ax.set_xticks(range(3)); ax.set_xticklabels(data.columns)
    ax.set_yticks(range(3)); ax.set_yticklabels(data.index)
    ax.set_title(title)
    for i in range(data.shape[0]):
        for j in range(data.shape[1]):
            val = data.values[i, j]
            ax.text(j, i, format(val, fmt) if not pd.isna(val) else 'n/a',
                    ha='center', va='center', color='white')
    plt.colorbar(im, ax=ax, shrink=0.8)

plt.tight_layout()
plt.show()

# boxplot grid: MECHA distribution per cell, 3 time zones side by side, 3 boxes each
fig, axes = plt.subplots(1, 3, figsize=(14, 4), sharey=True)
for ax, tz in zip(axes, ['early', 'mid', 'late']):
    data = [steps_df[(steps_df['time_zone']==tz) & (steps_df['zone']==z) & (steps_df['has_possession'])]['hull_area'].dropna()
            for z in ['win', 'draw', 'loss']]
    ax.boxplot(data, labels=['win', 'draw', 'loss'])
    ax.set_title(f'{tz}')
    ax.set_ylabel('hull area (MECHA)' if tz == 'early' else '')
plt.suptitle('MECHA distribution across all 9 (time_zone x score_zone) contexts')
plt.tight_layout()
plt.show()
```

```python
# %% [10] Outfield-only (players 1-4) formation shape across the 9 contexts
# GK confirmed pinned near own goal (player 0) from your earlier position-trail plot --
# excluded here since formation shape is an outfield concept.

OUTFIELD = [1, 2, 3, 4]
ZONES = ['win', 'draw', 'loss']
TIME_ZONES = ['early', 'mid', 'late']

def outfield_mean_positions(time_zone_target, zone_target):
    """Mean (x,y) per outfield player, pooled across all episodes/steps matching this context."""
    sums = {p: np.zeros(2) for p in OUTFIELD}
    counts = {p: 0 for p in OUTFIELD}

    for name, d in episodes.items():
        delta_score = d['score_left'] - d['score_right']
        zone = np.where(delta_score > 0, 'win', np.where(delta_score < 0, 'loss', 'draw'))
        t_norm = d['steps_left'] / 3001.0
        elapsed_min = (1 - t_norm) * 90.0
        time_zone = np.select([elapsed_min < 20, elapsed_min < 70], ['early', 'mid'], default='late')

        mask = (time_zone == time_zone_target) & (zone == zone_target)
        if mask.sum() == 0:
            continue
        for p in OUTFIELD:
            pts = d['left_team'][mask, p, :]
            sums[p] += pts.sum(axis=0)
            counts[p] += len(pts)

    means = {}
    for p in OUTFIELD:
        means[p] = sums[p] / counts[p] if counts[p] > 0 else None
    return means, sum(counts.values())


fig, axes = plt.subplots(3, 3, figsize=(12, 10), sharex=True, sharey=True)
colors = plt.cm.tab10(np.linspace(0, 1, 5))  # index by player number for consistent color across panels

for row, tz in enumerate(TIME_ZONES):
    for col, z in enumerate(ZONES):
        ax = axes[row, col]
        means, n_steps = outfield_mean_positions(tz, z)

        if n_steps == 0 or any(v is None for v in means.values()):
            ax.text(0.5, 0.5, 'no data', ha='center', va='center', transform=ax.transAxes)
            ax.set_facecolor('#eeeeee')
        else:
            pts = np.array([means[p] for p in OUTFIELD])  # shape (4, 2)

            # formation polygon: connect the 4 outfield mean positions in convex-hull order
            try:
                from scipy.spatial import ConvexHull
                hull = ConvexHull(pts)
                hull_pts = pts[hull.vertices]
                hull_pts = np.vstack([hull_pts, hull_pts[0]])  # close the polygon
                ax.plot(hull_pts[:, 0], hull_pts[:, 1], color='black', linewidth=1, alpha=0.6)
                ax.fill(hull_pts[:, 0], hull_pts[:, 1], color='gray', alpha=0.15)
            except Exception:
                pass  # degenerate (near-collinear) formation -- skip polygon, points still shown

            for p in OUTFIELD:
                ax.scatter(*means[p], color=colors[p], s=120, zorder=3, label=f'player {p}' if row==0 and col==0 else None)
                ax.annotate(str(p), means[p], textcoords="offset points", xytext=(5, 5), fontsize=9)

            centroid = pts.mean(axis=0)
            ax.scatter(*centroid, color='black', marker='x', s=60, zorder=4)

        ax.set_title(f'{tz} / {z}  (n={n_steps})', fontsize=10)
        ax.invert_yaxis()
        if row == 2:
            ax.set_xlabel('pitch x')
        if col == 0:
            ax.set_ylabel('pitch y')

axes[0, 0].legend(loc='upper left', fontsize=8, bbox_to_anchor=(1.0, 1.3), ncol=5)
fig.suptitle('Outfield formation shape (players 1-4, mean position) across 9 contexts\nblack x = centroid, shaded polygon = formation hull', y=1.04)
plt.tight_layout()
plt.show()
```

```python
# %% [11] Full per-context analytics battery: width, length, compactness (stretch index),
#          centroid, speed, possession, SAP -- for all 9 (time_zone x score_zone) contexts,
#          plus a time-series view for one representative episode.

from scipy.spatial.distance import pdist

OUTFIELD = [1, 2, 3, 4]
ZONES = ['win', 'draw', 'loss']
TIME_ZONES = ['early', 'mid', 'late']

def classify_all(d):
    delta_score = d['score_left'] - d['score_right']
    zone = np.where(delta_score > 0, 'win', np.where(delta_score < 0, 'loss', 'draw'))
    t_norm = d['steps_left'] / 3001.0
    elapsed_min = (1 - t_norm) * 90.0
    time_zone = np.select([elapsed_min < 20, elapsed_min < 70], ['early', 'mid'], default='late')
    return zone, time_zone

# ---- 1. Aggregate per-context metrics table ----
rows = []
for tz in TIME_ZONES:
    for z in ZONES:
        widths, lengths, compactness, centroids_x, centroids_y = [], [], [], [], []
        speeds, sprint_flags, poss_flags = [], [], []

        for name, d in episodes.items():
            zone, time_zone = classify_all(d)
            mask = (time_zone == tz) & (zone == z)
            if mask.sum() == 0:
                continue

            outfield_pos = d['left_team'][mask][:, OUTFIELD, :]      # (n, 4, 2)
            outfield_dir = d['left_team_direction'][mask][:, OUTFIELD, :]

            lengths.append(outfield_pos[:, :, 0].std(axis=1))         # spread along x, per step
            widths.append(outfield_pos[:, :, 1].std(axis=1))          # spread along y, per step
            centroids_x.append(outfield_pos[:, :, 0].mean(axis=1))
            centroids_y.append(outfield_pos[:, :, 1].mean(axis=1))
            speeds.append(np.linalg.norm(outfield_dir, axis=2).mean(axis=1))  # mean player speed, per step

            for t in range(outfield_pos.shape[0]):
                compactness.append(pdist(outfield_pos[t]).mean())     # avg pairwise distance = stretch index

            sprint_flags.append(d['sticky_actions'][mask, 8])
            poss_flags.append(d['has_possession'][mask])

        if not widths:
            rows.append({'time_zone': tz, 'zone': z, 'n_steps': 0})
            continue

        n_steps = sum(len(w) for w in widths)
        rows.append({
            'time_zone': tz, 'zone': z, 'n_steps': n_steps,
            'width_mean': np.concatenate(widths).mean(), 'width_std': np.concatenate(widths).std(),
            'length_mean': np.concatenate(lengths).mean(), 'length_std': np.concatenate(lengths).std(),
            'compactness_mean': np.mean(compactness), 'compactness_std': np.std(compactness),
            'centroid_x': np.concatenate(centroids_x).mean(), 'centroid_y': np.concatenate(centroids_y).mean(),
            'speed_mean': np.concatenate(speeds).mean(),
            'sap_pct': np.concatenate(sprint_flags).mean() * 100,
            'possession_pct': np.concatenate(poss_flags).mean() * 100,
        })

context_metrics_df = pd.DataFrame(rows)
pd.set_option('display.width', 160)
print(context_metrics_df.round(4))

# ---- 2. Heatmap grid for every metric ----
metrics_to_plot = ['width_mean', 'length_mean', 'compactness_mean', 'speed_mean', 'sap_pct', 'possession_pct']
fig, axes = plt.subplots(2, 3, figsize=(16, 8))
for ax, metric in zip(axes.ravel(), metrics_to_plot):
    pivot = context_metrics_df.pivot(index='time_zone', columns='zone', values=metric)
    pivot = pivot.reindex(index=TIME_ZONES, columns=ZONES)
    im = ax.imshow(pivot.values, cmap='viridis', aspect='auto')
    ax.set_xticks(range(3)); ax.set_xticklabels(ZONES)
    ax.set_yticks(range(3)); ax.set_yticklabels(TIME_ZONES)
    ax.set_title(metric)
    for i in range(3):
        for j in range(3):
            val = pivot.values[i, j]
            ax.text(j, i, f'{val:.2f}' if not pd.isna(val) else 'n/a', ha='center', va='center', color='white', fontsize=9)
    plt.colorbar(im, ax=ax, shrink=0.8)
fig.suptitle('Per-context analytics: 9 (time_zone x score_zone) contexts', y=1.02)
plt.tight_layout()
plt.show()

# ---- 3. Centroid position per context, plotted on a pitch outline ----
plt.figure(figsize=(7, 5))
markers = {'win': 'o', 'draw': 's', 'loss': '^'}
colors_tz = {'early': 'tab:blue', 'mid': 'tab:orange', 'late': 'tab:red'}
for _, row in context_metrics_df.iterrows():
    if row['n_steps'] == 0:
        continue
    plt.scatter(row['centroid_x'], row['centroid_y'], s=150,
                marker=markers[row['zone']], color=colors_tz[row['time_zone']],
                edgecolor='black', label=f"{row['time_zone']}/{row['zone']}")
plt.xlim(-1, 1); plt.ylim(-0.42, 0.42)
plt.gca().invert_yaxis()
plt.xlabel('pitch x'); plt.ylabel('pitch y')
plt.title('Team centroid by context (marker=score zone, color=time zone)')
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left', fontsize=8)
plt.tight_layout()
plt.show()

# ---- 4. Time-series view: width/length/compactness across one full representative episode ----
example_name = sorted(episodes.keys())[0]
d = episodes[example_name]
zone, time_zone = classify_all(d)
outfield_pos = d['left_team'][:, OUTFIELD, :]
width_ts = outfield_pos[:, :, 1].std(axis=1)
length_ts = outfield_pos[:, :, 0].std(axis=1)
compact_ts = np.array([pdist(outfield_pos[t]).mean() for t in range(len(d['steps_left']))])

fig, ax = plt.subplots(figsize=(14, 4))
ax.plot(width_ts, label='width (y-std)', alpha=0.8)
ax.plot(length_ts, label='length (x-std)', alpha=0.8)
ax.plot(compact_ts, label='compactness (mean pairwise dist)', alpha=0.8)

# shade by score zone
zone_colors = {'win': 'lightgreen', 'draw': 'lightgray', 'loss': 'lightcoral'}
prev = zone[0]; start = 0
for i in range(1, len(zone) + 1):
    if i == len(zone) or zone[i] != prev:
        ax.axvspan(start, i, color=zone_colors[prev], alpha=0.15)
        if i < len(zone):
            start = i; prev = zone[i]

ax.set_xlabel('step'); ax.set_ylabel('value')
ax.set_title(f'Formation shape time series -- {example_name}\n(background shading: green=win, gray=draw, red=loss)')
ax.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

```python
# %% [12] Full per-player analysis (positions, movement, pitch-thirds, involvement) across all 9 contexts
OUTFIELD_ALL = [0, 1, 2, 3, 4]   # include GK too, for completeness
ZONES = ['win', 'draw', 'loss']
TIME_ZONES = ['early', 'mid', 'late']
THIRDS = {'defensive': (-1.0, -0.33), 'middle': (-0.33, 0.33), 'attacking': (0.33, 1.0)}

def classify_all(d):
    delta_score = d['score_left'] - d['score_right']
    zone = np.where(delta_score > 0, 'win', np.where(delta_score < 0, 'loss', 'draw'))
    t_norm = d['steps_left'] / 3001.0
    elapsed_min = (1 - t_norm) * 90.0
    time_zone = np.select([elapsed_min < 20, elapsed_min < 70], ['early', 'mid'], default='late')
    return zone, time_zone

player_rows = []
for p in OUTFIELD_ALL:
    for tz in TIME_ZONES:
        for z in ZONES:
            xs, ys, speeds, dists, involved = [], [], [], [], []
            n_steps = 0
            for name, d in episodes.items():
                zone, time_zone = classify_all(d)
                mask = (time_zone == tz) & (zone == z)
                if mask.sum() == 0:
                    continue
                pos = d['left_team'][mask, p, :]
                direction = d['left_team_direction'][mask, p, :]
                speed = np.linalg.norm(direction, axis=1)

                xs.append(pos[:, 0]); ys.append(pos[:, 1]); speeds.append(speed)
                dists.append(speed.sum())  # approx distance covered = sum of per-step speed

                # involvement: this player owns the ball -- depends on unverified ball_owned_player indexing
                bot = d['ball_owned_team'][mask]; bop = d['ball_owned_player'][mask]
                involved.append(((bot == 0) & (bop == p)).sum())
                n_steps += mask.sum()

            if n_steps == 0:
                player_rows.append({'player': p, 'time_zone': tz, 'zone': z, 'n_steps': 0})
                continue

            all_x, all_y, all_speed = np.concatenate(xs), np.concatenate(ys), np.concatenate(speeds)
            third_pct = {
                name: float(((all_x >= lo) & (all_x < hi)).mean() * 100)
                for name, (lo, hi) in THIRDS.items()
            }
            player_rows.append({
                'player': p, 'time_zone': tz, 'zone': z, 'n_steps': int(n_steps),
                'avg_x': all_x.mean(), 'avg_y': all_y.mean(),
                'std_x': all_x.std(), 'std_y': all_y.std(),
                'avg_speed': all_speed.mean(), 'total_distance': sum(dists),
                'involvement_pct': sum(involved) / n_steps * 100,
                **{f'pct_{k}_third': v for k, v in third_pct.items()},
            })

player_df = pd.DataFrame(player_rows)
pd.set_option('display.width', 200)
print(player_df.round(3))

# ---- Visualization 1: distance covered per player, grouped by time zone (role indicator) ----
fig, ax = plt.subplots(figsize=(10, 5))
width = 0.25
x = np.arange(5)
for i, tz in enumerate(TIME_ZONES):
    sub = player_df[player_df['time_zone'] == tz].groupby('player')['total_distance'].sum()
    sub = sub.reindex(OUTFIELD_ALL, fill_value=0)
    ax.bar(x + i*width, sub.values, width, label=tz)
ax.set_xticks(x + width); ax.set_xticklabels([f'player {p}' for p in OUTFIELD_ALL])
ax.set_ylabel('total distance covered (sum |direction|)')
ax.set_title('Distance covered per player, by time zone')
ax.legend()
plt.tight_layout(); plt.show()

# ---- Visualization 2: pitch-third occupancy per player (stacked bar -- role profile) ----
overall_thirds = player_df.groupby('player')[['pct_defensive_third', 'pct_middle_third', 'pct_attacking_third']].mean()
overall_thirds = overall_thirds.reindex(OUTFIELD_ALL)
fig, ax = plt.subplots(figsize=(8, 5))
bottom = np.zeros(5)
for col, color in zip(['pct_defensive_third', 'pct_middle_third', 'pct_attacking_third'],
                       ['tab:blue', 'tab:gray', 'tab:red']):
    ax.bar([f'player {p}' for p in OUTFIELD_ALL], overall_thirds[col], bottom=bottom, label=col.replace('pct_','').replace('_third',''), color=color)
    bottom += overall_thirds[col].values
ax.set_ylabel('% of time in third')
ax.set_title('Pitch-third occupancy per player (role profile, averaged across all contexts)')
ax.legend()
plt.tight_layout(); plt.show()

# ---- Visualization 3: involvement (ball ownership) heatmap, player x context ----
fig, ax = plt.subplots(figsize=(8, 5))
inv_pivot = player_df.pivot_table(index='player', columns=['time_zone', 'zone'], values='involvement_pct')
inv_pivot = inv_pivot.reindex(index=OUTFIELD_ALL)
im = ax.imshow(inv_pivot.values, cmap='viridis', aspect='auto')
ax.set_yticks(range(5)); ax.set_yticklabels([f'player {p}' for p in OUTFIELD_ALL])
ax.set_xticks(range(len(inv_pivot.columns)))
ax.set_xticklabels([f'{tz}\n{z}' for tz, z in inv_pivot.columns], fontsize=8)
ax.set_title('Ball involvement % per player, across 9 contexts\n(CAVEAT: depends on unverified ball_owned_player indexing)')
plt.colorbar(im, ax=ax, label='involvement %')
plt.tight_layout(); plt.show()

# ---- Visualization 4: speed distribution per player (boxplot) ----
fig, ax = plt.subplots(figsize=(8, 5))
speed_data = []
for p in OUTFIELD_ALL:
    all_speeds = []
    for name, d in episodes.items():
        all_speeds.append(np.linalg.norm(d['left_team_direction'][:, p, :], axis=1))
    speed_data.append(np.concatenate(all_speeds))
ax.boxplot(speed_data, labels=[f'player {p}' for p in OUTFIELD_ALL])
ax.set_ylabel('speed (|direction|)')
ax.set_title('Speed distribution per player, all episodes/contexts pooled')
plt.tight_layout(); plt.show()
```

```python
# %% [13] Mean positions per context (formation snapshots) + pass direction (forward/backward) per player

OUTFIELD_ALL = [0, 1, 2, 3, 4]
ZONES = ['win', 'draw', 'loss']
TIME_ZONES = ['early', 'mid', 'late']
PASS_ACTIONS = {9: 'long_pass', 10: 'high_pass', 11: 'short_pass'}

def classify_all(d):
    delta_score = d['score_left'] - d['score_right']
    zone = np.where(delta_score > 0, 'win', np.where(delta_score < 0, 'loss', 'draw'))
    t_norm = d['steps_left'] / 3001.0
    elapsed_min = (1 - t_norm) * 90.0
    time_zone = np.select([elapsed_min < 20, elapsed_min < 70], ['early', 'mid'], default='late')
    return zone, time_zone

# ---- 1. Mean position per player, per context ----
mean_pos_rows = []
for tz in TIME_ZONES:
    for z in ZONES:
        pos_by_player = {p: [] for p in OUTFIELD_ALL}
        n_steps = 0
        for name, d in episodes.items():
            zone, time_zone = classify_all(d)
            mask = (time_zone == tz) & (zone == z)
            if mask.sum() == 0:
                continue
            n_steps += mask.sum()
            for p in OUTFIELD_ALL:
                pos_by_player[p].append(d['left_team'][mask, p, :])
        for p in OUTFIELD_ALL:
            if not pos_by_player[p]:
                continue
            pts = np.concatenate(pos_by_player[p])
            mean_pos_rows.append({'player': p, 'time_zone': tz, 'zone': z, 'n_steps': int(n_steps),
                                   'mean_x': pts[:, 0].mean(), 'mean_y': pts[:, 1].mean()})

mean_pos_df = pd.DataFrame(mean_pos_rows)
print(mean_pos_df.round(3))

# ---- 2. Formation snapshot: mean position per player, 3x3 grid of contexts ----
colors = plt.cm.tab10(np.linspace(0, 1, 5))
fig, axes = plt.subplots(3, 3, figsize=(12, 10), sharex=True, sharey=True)
for row, tz in enumerate(TIME_ZONES):
    for col, z in enumerate(ZONES):
        ax = axes[row, col]
        sub = mean_pos_df[(mean_pos_df['time_zone'] == tz) & (mean_pos_df['zone'] == z)]
        if sub.empty:
            ax.text(0.5, 0.5, 'no data', ha='center', va='center', transform=ax.transAxes)
        else:
            for _, r in sub.iterrows():
                ax.scatter(r['mean_x'], r['mean_y'], s=140, color=colors[int(r['player'])], zorder=3)
                ax.annotate(str(int(r['player'])), (r['mean_x'], r['mean_y']),
                            textcoords="offset points", xytext=(5, 5), fontsize=9)
            n = int(sub['n_steps'].iloc[0])
            ax.set_title(f'{tz} / {z}  (n={n})', fontsize=10)
        ax.set_xlim(-1, 1); ax.set_ylim(-0.42, 0.42); ax.invert_yaxis()
        if row == 2: ax.set_xlabel('pitch x')
        if col == 0: ax.set_ylabel('pitch y')
fig.suptitle('Mean position snapshot per player, across 9 contexts', y=1.02)
plt.tight_layout(); plt.show()

# ---- 3. Pass direction (forward/backward) per player ----
pass_rows = []
for name, d in episodes.items():
    zone, time_zone = classify_all(d)
    for step in range(len(d['steps_left']) - 1):
        act = d['action_captured'][step]
        if act not in PASS_ACTIONS:
            continue
        if d['ball_owned_team'][step] != 0:
            continue  # shouldn't happen given attack-mode gating, but guard anyway
        passer = int(d['ball_owned_player'][step])
        ball_dir_x = d['ball_direction'][step, 0]
        direction = 'forward' if ball_dir_x > 0 else ('backward' if ball_dir_x < 0 else 'neutral')
        pass_rows.append({
            'source_file': name, 'step': step, 'player': passer,
            'pass_type': PASS_ACTIONS[act], 'direction': direction,
            'zone': zone[step], 'time_zone': time_zone[step],
        })

pass_df = pd.DataFrame(pass_rows)
print(f"\nTotal pass events captured: {len(pass_df)}")
print(pass_df['direction'].value_counts())
print()

# per-player forward/backward counts
pass_summary = pass_df.groupby(['player', 'direction']).size().unstack(fill_value=0)
pass_summary = pass_summary.reindex(index=OUTFIELD_ALL, columns=['forward', 'backward', 'neutral'], fill_value=0)
print(pass_summary)

# visualization: stacked bar, forward/backward per player
fig, ax = plt.subplots(figsize=(8, 5))
bottom = np.zeros(5)
for col, color in zip(['forward', 'backward', 'neutral'], ['tab:green', 'tab:red', 'tab:gray']):
    vals = pass_summary[col].values if col in pass_summary.columns else np.zeros(5)
    ax.bar([f'player {p}' for p in OUTFIELD_ALL], vals, bottom=bottom, label=col, color=color)
    bottom += vals
ax.set_ylabel('pass count')
ax.set_title('Pass direction per player (forward = toward opponent goal, +x)')
ax.legend()
plt.tight_layout(); plt.show()

# per-context forward-pass-rate heatmap (does the team pass more forward when winning vs losing?)
fwd_rate = pass_df.groupby(['time_zone', 'zone']).apply(
    lambda g: (g['direction'] == 'forward').mean() * 100 if len(g) > 0 else np.nan
).unstack().reindex(index=TIME_ZONES, columns=ZONES)

plt.figure(figsize=(6, 4))
im = plt.imshow(fwd_rate.values, cmap='RdYlGn', aspect='auto', vmin=0, vmax=100)
plt.xticks(range(3), ZONES); plt.yticks(range(3), TIME_ZONES)
for i in range(3):
    for j in range(3):
        v = fwd_rate.values[i, j]
        plt.text(j, i, f'{v:.0f}%' if not pd.isna(v) else 'n/a', ha='center', va='center')
plt.colorbar(im, label='% passes forward')
plt.title('Forward-pass rate by context')
plt.tight_layout(); plt.show()
```

```python
# %% [14] Possession-based analytics: team on/off, player-level on/off-ball, transitions

OUTFIELD_ALL = [0, 1, 2, 3, 4]
ZONES = ['win', 'draw', 'loss']
TIME_ZONES = ['early', 'mid', 'late']

def classify_all(d):
    delta_score = d['score_left'] - d['score_right']
    zone = np.where(delta_score > 0, 'win', np.where(delta_score < 0, 'loss', 'draw'))
    t_norm = d['steps_left'] / 3001.0
    elapsed_min = (1 - t_norm) * 90.0
    time_zone = np.select([elapsed_min < 20, elapsed_min < 70], ['early', 'mid'], default='late')
    return zone, time_zone

# ---- 1. Team shape: on-possession vs off-possession, width/length/compactness ----
from scipy.spatial.distance import pdist
rows = []
for poss_state in [True, False]:
    widths, lengths, compact = [], [], []
    for name, d in episodes.items():
        mask = d['has_possession'] == poss_state
        if mask.sum() == 0:
            continue
        pos = d['left_team'][mask][:, [1,2,3,4], :]  # outfield only
        widths.append(pos[:, :, 1].std(axis=1))
        lengths.append(pos[:, :, 0].std(axis=1))
        for t in range(pos.shape[0]):
            compact.append(pdist(pos[t]).mean())
    rows.append({
        'possession': 'on' if poss_state else 'off',
        'width_mean': np.concatenate(widths).mean(), 'length_mean': np.concatenate(lengths).mean(),
        'compactness_mean': np.mean(compact),
    })
poss_shape_df = pd.DataFrame(rows)
print(poss_shape_df.round(4))

fig, ax = plt.subplots(figsize=(7,4))
poss_shape_df.set_index('possession')[['width_mean','length_mean','compactness_mean']].plot.bar(ax=ax)
ax.set_title('Team formation shape: on-possession vs off-possession')
ax.set_ylabel('value')
plt.xticks(rotation=0)
plt.tight_layout(); plt.show()

# ---- 2. Player-level on-ball vs off-ball: mean position, speed ----
onball_rows = []
for p in OUTFIELD_ALL:
    for state in ['on_ball', 'off_ball']:
        xs, ys, speeds = [], [], []
        for name, d in episodes.items():
            if state == 'on_ball':
                mask = (d['ball_owned_team'] == 0) & (d['ball_owned_player'] == p)
            else:
                mask = ~((d['ball_owned_team'] == 0) & (d['ball_owned_player'] == p))
            if mask.sum() == 0:
                continue
            xs.append(d['left_team'][mask, p, 0]); ys.append(d['left_team'][mask, p, 1])
            speeds.append(np.linalg.norm(d['left_team_direction'][mask, p, :], axis=1))
        if not xs:
            continue
        onball_rows.append({
            'player': p, 'state': state,
            'mean_x': np.concatenate(xs).mean(), 'mean_y': np.concatenate(ys).mean(),
            'mean_speed': np.concatenate(speeds).mean(), 'n_steps': sum(len(x) for x in xs),
        })
onball_df = pd.DataFrame(onball_rows)
print(onball_df.round(3))

fig, ax = plt.subplots(figsize=(8,5))
width = 0.35
x = np.arange(5)
on_speed = onball_df[onball_df['state']=='on_ball'].set_index('player')['mean_speed'].reindex(OUTFIELD_ALL, fill_value=0)
off_speed = onball_df[onball_df['state']=='off_ball'].set_index('player')['mean_speed'].reindex(OUTFIELD_ALL, fill_value=0)
ax.bar(x - width/2, on_speed, width, label='on-ball', color='tab:orange')
ax.bar(x + width/2, off_speed, width, label='off-ball', color='tab:blue')
ax.set_xticks(x); ax.set_xticklabels([f'player {p}' for p in OUTFIELD_ALL])
ax.set_ylabel('mean speed')
ax.set_title('On-ball vs off-ball speed, per player (CAVEAT: rests on ball_owned_player indexing)')
ax.legend()
plt.tight_layout(); plt.show()

# ---- 3. Possession % by context (win/draw/loss x early/mid/late) ----
poss_rows = []
for tz in TIME_ZONES:
    for z in ZONES:
        vals = []
        for name, d in episodes.items():
            zone, time_zone = classify_all(d)
            mask = (time_zone == tz) & (zone == z)
            if mask.sum() == 0:
                continue
            vals.append(d['has_possession'][mask])
        if vals:
            poss_rows.append({'time_zone': tz, 'zone': z, 'possession_pct': np.concatenate(vals).mean()*100})
poss_ctx_df = pd.DataFrame(poss_rows)
poss_pivot = poss_ctx_df.pivot(index='time_zone', columns='zone', values='possession_pct').reindex(index=TIME_ZONES, columns=ZONES)

plt.figure(figsize=(6,4))
im = plt.imshow(poss_pivot.values, cmap='RdYlGn', aspect='auto', vmin=0, vmax=100)
plt.xticks(range(3), ZONES); plt.yticks(range(3), TIME_ZONES)
for i in range(3):
    for j in range(3):
        v = poss_pivot.values[i,j]
        plt.text(j, i, f'{v:.0f}%' if not pd.isna(v) else 'n/a', ha='center', va='center')
plt.colorbar(im, label='possession %')
plt.title('Possession % by context')
plt.tight_layout(); plt.show()

# ---- 4. Possession streaks: how long do spells last? ----
streak_lengths = []
for name, d in episodes.items():
    poss = d['has_possession']
    changes = np.where(np.diff(poss.astype(int)) != 0)[0]
    bounds = np.concatenate([[0], changes+1, [len(poss)]])
    for i in range(len(bounds)-1):
        streak_lengths.append(bounds[i+1] - bounds[i])
streak_lengths = np.array(streak_lengths)

plt.figure(figsize=(8,4))
plt.hist(streak_lengths, bins=50)
plt.xlabel('possession spell length (steps)')
plt.ylabel('count')
plt.title(f'Distribution of possession spell lengths, all episodes\nmean={streak_lengths.mean():.1f} steps, median={np.median(streak_lengths):.0f} steps')
plt.tight_layout(); plt.show()

# ---- 5. Transition moments: team shape in the 20 steps right after winning/losing the ball ----
WINDOW = 20
gain_widths, loss_widths = [], []
for name, d in episodes.items():
    poss = d['has_possession'].astype(int)
    diffs = np.diff(poss)
    gains = np.where(diffs == 1)[0] + 1   # step where possession just started
    losses = np.where(diffs == -1)[0] + 1 # step where possession just ended
    outfield_pos = d['left_team'][:, [1,2,3,4], :]
    for g in gains:
        if g + WINDOW < len(d['steps_left']):
            gain_widths.append(outfield_pos[g:g+WINDOW, :, 1].std(axis=1).mean())
    for l in losses:
        if l + WINDOW < len(d['steps_left']):
            loss_widths.append(outfield_pos[l:l+WINDOW, :, 1].std(axis=1).mean())

print(f"\nTransition moments: {len(gain_widths)} possession gains, {len(loss_widths)} possession losses")
print(f"Team width, {WINDOW} steps after gaining possession: mean={np.mean(gain_widths):.4f}")
print(f"Team width, {WINDOW} steps after losing possession:  mean={np.mean(loss_widths):.4f}")

plt.figure(figsize=(6,4))
plt.boxplot([gain_widths, loss_widths], labels=['after gaining', 'after losing'])
plt.ylabel('team width (y-std)')
plt.title(f'Team width in the {WINDOW} steps after a possession transition')
plt.tight_layout(); plt.show()
```

```python
# %% [15] World-class visualizations: pitch control (Voronoi), player radar chart, movement-direction rose

from scipy.spatial import Voronoi
import matplotlib.patches as patches

OUTFIELD_ALL = [0, 1, 2, 3, 4]
PITCH_X, PITCH_Y = (-1.0, 1.0), (-0.42, 0.42)

# ============================================================
# 1. VORONOI PITCH CONTROL -- space dominance at one frozen frame
#    Answers: "who controls which part of the pitch, right now"
#    -- exactly the "frozen episode" idea from earlier in this project.
# ============================================================
def plot_voronoi_frame(d, step, ax=None):
    left = d['left_team'][step]    # (5, 2)
    right = d['right_team'][step]  # (5, 2)
    ball = d['ball'][step, :2]
    all_pts = np.vstack([left, right])

    # mirror points far outside the pitch so Voronoi cells near the edge
    # close properly instead of extending to infinity
    mirrors = []
    for (x, y) in all_pts:
        mirrors.append([x, 2*PITCH_Y[1] - y + 5])
        mirrors.append([x, 2*PITCH_Y[0] - y - 5])
        mirrors.append([2*PITCH_X[1] - x + 5, y])
        mirrors.append([2*PITCH_X[0] - x - 5, y])
    vor_pts = np.vstack([all_pts, mirrors])
    vor = Voronoi(vor_pts)

    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 5))

    for i in range(10):  # only the 10 real players, ignore mirror regions
        region_idx = vor.point_region[i]
        region = vor.regions[region_idx]
        if -1 in region or len(region) == 0:
            continue
        polygon = [vor.vertices[v] for v in region]
        color = 'tab:blue' if i < 5 else 'tab:red'
        ax.fill(*zip(*polygon), alpha=0.25, color=color, edgecolor='black', linewidth=0.5)

    ax.scatter(left[:, 0], left[:, 1], color='tab:blue', s=100, zorder=3, label='left team')
    ax.scatter(right[:, 0], right[:, 1], color='tab:red', s=100, zorder=3, label='right team')
    ax.scatter(*ball, color='black', marker='o', s=40, zorder=4, label='ball')
    for i, (x, y) in enumerate(left):
        ax.annotate(str(i), (x, y), textcoords="offset points", xytext=(4,4), fontsize=8)

    ax.set_xlim(*PITCH_X); ax.set_ylim(*PITCH_Y)
    ax.invert_yaxis()
    ax.set_title(f'Pitch control -- step {step}')
    ax.legend(loc='upper right', fontsize=8)
    return ax

example_name = sorted(episodes.keys())[0]
d = episodes[example_name]
mid_step = len(d['steps_left']) // 2
plot_voronoi_frame(d, mid_step)
plt.tight_layout()
plt.show()

# ============================================================
# 2. PLAYER RADAR CHART -- multi-metric profile comparison, all 5 players
#    Answers: "what kind of player is each index, in one glance"
# ============================================================
metrics_names = ['distance', 'avg_speed', 'sap_%', 'attack_third_%', 'involvement_%']
player_metrics = {p: {} for p in OUTFIELD_ALL}

for p in OUTFIELD_ALL:
    dists, speeds, sprints, attack_pct, invs = [], [], [], [], []
    for name, dd in episodes.items():
        direction = dd['left_team_direction'][:, p, :]
        speed = np.linalg.norm(direction, axis=1)
        dists.append(speed.sum())
        speeds.append(speed.mean())
        sprints.append(dd['sticky_actions'][:, 8].mean())  # note: team-level sprint signal, caveat as before
        attack_pct.append((dd['left_team'][:, p, 0] > 0.33).mean())
        invs.append(((dd['ball_owned_team']==0) & (dd['ball_owned_player']==p)).mean())
    player_metrics[p] = {
        'distance': np.mean(dists), 'avg_speed': np.mean(speeds),
        'sap_%': np.mean(sprints)*100, 'attack_third_%': np.mean(attack_pct)*100,
        'involvement_%': np.mean(invs)*100,
    }

metrics_matrix = np.array([[player_metrics[p][m] for m in metrics_names] for p in OUTFIELD_ALL])
# min-max normalize each metric column across players, so radar shapes are comparable
norm_matrix = (metrics_matrix - metrics_matrix.min(axis=0)) / (metrics_matrix.max(axis=0) - metrics_matrix.min(axis=0) + 1e-9)

angles = np.linspace(0, 2*np.pi, len(metrics_names), endpoint=False).tolist()
angles += angles[:1]

fig, axes = plt.subplots(1, 5, figsize=(22, 4.5), subplot_kw=dict(polar=True))
colors = plt.cm.tab10(np.linspace(0, 1, 5))
for idx, p in enumerate(OUTFIELD_ALL):
    ax = axes[idx]
    vals = norm_matrix[idx].tolist() + [norm_matrix[idx][0]]
    ax.plot(angles, vals, color=colors[p], linewidth=2)
    ax.fill(angles, vals, color=colors[p], alpha=0.25)
    ax.set_xticks(angles[:-1]); ax.set_xticklabels(metrics_names, fontsize=8)
    ax.set_yticks([]); ax.set_ylim(0, 1)
    ax.set_title(f'player {p}', fontsize=11)
plt.suptitle('Player profile radar (normalized 0-1 across the 5 players)', y=1.05)
plt.tight_layout()
plt.show()

# ============================================================
# 3. MOVEMENT-DIRECTION ROSE -- polar histogram of heading, per player
#    Answers: "which direction does this player actually move in most"
# ============================================================
fig, axes = plt.subplots(1, 5, figsize=(20, 4.5), subplot_kw=dict(polar=True))
for idx, p in enumerate(OUTFIELD_ALL):
    all_angles, all_speeds = [], []
    for name, dd in episodes.items():
        direction = dd['left_team_direction'][:, p, :]
        speed = np.linalg.norm(direction, axis=1)
        moving = speed > 1e-4
        angle = np.arctan2(direction[moving, 1], direction[moving, 0])
        all_angles.append(angle)
        all_speeds.append(speed[moving])
    all_angles = np.concatenate(all_angles)
    all_speeds = np.concatenate(all_speeds)

    bins = np.linspace(-np.pi, np.pi, 17)
    hist, _ = np.histogram(all_angles, bins=bins, weights=all_speeds)
    width = bins[1] - bins[0]
    ax = axes[idx]
    ax.bar(bins[:-1], hist, width=width, color=colors[p], alpha=0.7, align='edge')
    ax.set_title(f'player {p}', fontsize=11)
    ax.set_yticklabels([])
plt.suptitle('Movement-direction rose per player (speed-weighted, 0°=toward opponent goal)', y=1.05)
plt.tight_layout()
plt.show()
```

```python
# %% [16] Overlaid radar, pass network (inferred), movement flow field, lollipop ranking

OUTFIELD_ALL = [0, 1, 2, 3, 4]
ZONES = ['win', 'draw', 'loss']

def classify_all(d):
    delta_score = d['score_left'] - d['score_right']
    zone = np.where(delta_score > 0, 'win', np.where(delta_score < 0, 'loss', 'draw'))
    return zone

# ============================================================
# A. OVERLAID RADAR -- team profile compared across win/draw/loss, one chart, 3 polygons
# ============================================================
from scipy.spatial.distance import pdist

metrics_names = ['width', 'length', 'compactness', 'possession_%', 'sap_%']
zone_metrics = {}
for z in ZONES:
    widths, lengths, compact, poss, sprint = [], [], [], [], []
    for name, d in episodes.items():
        zone = classify_all(d)
        mask = zone == z
        if mask.sum() == 0:
            continue
        pos = d['left_team'][mask][:, [1,2,3,4], :]
        widths.append(pos[:,:,1].std(axis=1)); lengths.append(pos[:,:,0].std(axis=1))
        for t in range(pos.shape[0]):
            compact.append(pdist(pos[t]).mean())
        poss.append(d['has_possession'][mask]); sprint.append(d['sticky_actions'][mask, 8])
    if not widths:
        continue
    zone_metrics[z] = [
        np.concatenate(widths).mean(), np.concatenate(lengths).mean(), np.mean(compact),
        np.concatenate(poss).mean()*100, np.concatenate(sprint).mean()*100,
    ]

matrix = np.array(list(zone_metrics.values()))
norm = (matrix - matrix.min(axis=0)) / (matrix.max(axis=0) - matrix.min(axis=0) + 1e-9)
angles = np.linspace(0, 2*np.pi, len(metrics_names), endpoint=False).tolist(); angles += angles[:1]

fig = plt.figure(figsize=(6,6))
ax = fig.add_subplot(111, polar=True)
zone_colors = {'win':'tab:green', 'draw':'tab:gray', 'loss':'tab:red'}
for z, vals in zip(zone_metrics.keys(), norm):
    v = vals.tolist() + [vals[0]]
    ax.plot(angles, v, color=zone_colors[z], linewidth=2, label=z)
    ax.fill(angles, v, color=zone_colors[z], alpha=0.15)
ax.set_xticks(angles[:-1]); ax.set_xticklabels(metrics_names)
ax.set_yticks([])
ax.set_title('Team profile radar, overlaid by score zone')
ax.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))
plt.tight_layout(); plt.show()

# ============================================================
# B. PASS NETWORK -- inferred receiver via next-possession-change tracking
#    NOT ground truth: receiver is inferred as whichever left-team player
#    next gains possession after a pass, within a short window, without the
#    right team gaining it first. Flagged as inferred throughout.
# ============================================================
WINDOW = 40
pass_edges = {}  # (passer, receiver) -> count
node_positions = {p: [] for p in OUTFIELD_ALL}

for name, d in episodes.items():
    n = len(d['steps_left'])
    for t in range(n):
        node_positions[int(d['ball_owned_player'][t])].append(d['left_team'][t, int(d['ball_owned_player'][t])]) \
            if d['ball_owned_team'][t] == 0 and 0 <= d['ball_owned_player'][t] < 5 else None

        if d['action_captured'][t] not in (9, 10, 11):  # pass actions only
            continue
        if d['ball_owned_team'][t] != 0:
            continue
        passer = int(d['ball_owned_player'][t])

        for look in range(t+1, min(t+WINDOW, n)):
            if d['ball_owned_team'][look] == 1:
                break  # intercepted -- don't count as a completed pass
            if d['ball_owned_team'][look] == 0 and int(d['ball_owned_player'][look]) != passer:
                receiver = int(d['ball_owned_player'][look])
                pass_edges[(passer, receiver)] = pass_edges.get((passer, receiver), 0) + 1
                break

node_mean_pos = {p: np.mean(node_positions[p], axis=0) if node_positions[p] else np.array([0,0])
                  for p in OUTFIELD_ALL}

print(f"Inferred completed passes: {sum(pass_edges.values())}")
print("Top connections:", sorted(pass_edges.items(), key=lambda x: -x[1])[:10])

fig, ax = plt.subplots(figsize=(8,5))
max_count = max(pass_edges.values()) if pass_edges else 1
for (a, b), count in pass_edges.items():
    p1, p2 = node_mean_pos[a], node_mean_pos[b]
    ax.plot([p1[0], p2[0]], [p1[1], p2[1]], color='gray', alpha=0.6, linewidth=1 + 4*count/max_count)
for p in OUTFIELD_ALL:
    ax.scatter(*node_mean_pos[p], s=250, color=colors[p] if 'colors' in dir() else 'tab:blue', zorder=3, edgecolor='black')
    ax.annotate(str(p), node_mean_pos[p], textcoords="offset points", xytext=(6,6), fontsize=10)
ax.set_xlim(-1,1); ax.set_ylim(-0.42,0.42); ax.invert_yaxis()
ax.set_title(f'Inferred pass network (edge width = frequency, n={sum(pass_edges.values())} passes)')
plt.tight_layout(); plt.show()

# ============================================================
# C. MOVEMENT FLOW FIELD -- vector field of average heading per pitch cell (all players pooled)
# ============================================================
GRID_X, GRID_Y = 12, 8
x_edges = np.linspace(-1, 1, GRID_X+1)
y_edges = np.linspace(-0.42, 0.42, GRID_Y+1)
x_centers = (x_edges[:-1] + x_edges[1:]) / 2
y_centers = (y_edges[:-1] + y_edges[1:]) / 2

sum_dx = np.zeros((GRID_Y, GRID_X)); sum_dy = np.zeros((GRID_Y, GRID_X)); count = np.zeros((GRID_Y, GRID_X))
for name, d in episodes.items():
    for p in OUTFIELD_ALL:
        xs = d['left_team'][:, p, 0]; ys = d['left_team'][:, p, 1]
        dxs = d['left_team_direction'][:, p, 0]; dys = d['left_team_direction'][:, p, 1]
        xi = np.clip(np.digitize(xs, x_edges) - 1, 0, GRID_X-1)
        yi = np.clip(np.digitize(ys, y_edges) - 1, 0, GRID_Y-1)
        np.add.at(sum_dx, (yi, xi), dxs); np.add.at(sum_dy, (yi, xi), dys); np.add.at(count, (yi, xi), 1)

with np.errstate(invalid='ignore'):
    mean_dx = np.where(count > 0, sum_dx / count, 0)
    mean_dy = np.where(count > 0, sum_dy / count, 0)

fig, ax = plt.subplots(figsize=(9,5))
Xc, Yc = np.meshgrid(x_centers, y_centers)
ax.quiver(Xc, Yc, mean_dx, mean_dy, count, cmap='viridis', scale=2)
ax.set_xlim(-1,1); ax.set_ylim(-0.42,0.42); ax.invert_yaxis()
ax.set_title('Movement flow field -- average heading per pitch cell, all players pooled')
plt.tight_layout(); plt.show()

# ============================================================
# D. LOLLIPOP CHART -- clean ranked comparison, distance covered per player
# ============================================================
dist_by_player = {}
for p in OUTFIELD_ALL:
    total = sum(np.linalg.norm(d['left_team_direction'][:, p, :], axis=1).sum() for d in episodes.values())
    dist_by_player[p] = total

ranked = sorted(dist_by_player.items(), key=lambda x: x[1])
players_r, dists_r = zip(*ranked)

fig, ax = plt.subplots(figsize=(7,4))
ax.hlines(y=[f'player {p}' for p in players_r], xmin=0, xmax=dists_r, color='gray', linewidth=2)
ax.plot(dists_r, [f'player {p}' for p in players_r], 'o', color='tab:blue', markersize=10)
ax.set_xlabel('total distance covered (sum |direction|), all episodes')
ax.set_title('Distance covered per player -- ranked')
plt.tight_layout(); plt.show()
```
